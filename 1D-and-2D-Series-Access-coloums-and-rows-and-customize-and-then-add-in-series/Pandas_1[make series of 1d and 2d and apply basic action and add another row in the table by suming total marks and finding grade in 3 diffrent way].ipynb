# DataFrame
   - data frame contain more then 1 coloum and it is called 2d
   - data frame give clean ui when we load date through pd.dataframe()

```python
s3 = pd.DataFrame({"name":["hazzan","ali","ilyas","ilaqat","zubair","abdullah","saad","azaan","haider","haya"],
             "pandas":[67,89,90,87,65,45,34,78,90,99],
             "statistics":[89,78,67,89,56,45,36,87,90,90],
             "python":[56,67,87,97,35,67,46,75,75,65],
             "reputation":["good","bad","normal","good","bad","normal","good","bad","normal","bad"],
            })

s3
```

```python
#mrthod 1 to add
s3['Total_Mark'] = s3['statistics']+s3['python']+s3["pandas"]
#method 2 to add
s3['Total_Mark_2'] = s3.statistics+s3.python+s3.pandas
```

```python
s3
```

```python
s3["percentage"] = round(s3.Total_Mark_2/300*100,2)
```

```python
s3
```

```python
#method one to add Grade in data frame
#List Comprehension simpel formula for list comprehension [what_if true if condition else another contition/value for dash in dah.dash]
s3["Grade_1"] = ["A+" if per>=90 
                else 
                   "A" if per>=80 
                else
                    "B" if per>=70 
                else
                   "C" if per>=40 
                else
                   "D" if per >=33
                else "F"
                             for per in s3.percentage]
```

```python
s3
```

```python
#mrthod two to add grade in data frame 
#lambda function is a anonymus function which have no name it create in one line and also destry after finish
s3['Grade_2'] = s3.percentage.apply(lambda per: "A+" if per>=90 else "A" if per>=80 else "B" if per>=70 else "C" if per>=50 else "D" if per>=33 else "F")
```

```python
s3
```

```python
#mrthod three to add grade in data frame
#which is create a outline function and give there name series

def grade_calc(per):
    if per>= 90:
        return "A+"
    elif per>= 80:
        return "A"
    elif per >= 70:
        return "B"
    elif per>= 50:
        return "C"
    elif per>= 33:
        return "D"
    else:
        "F"
```

```python
s3["Grade_3"] = s3.percentage.apply(grade_calc)
```
